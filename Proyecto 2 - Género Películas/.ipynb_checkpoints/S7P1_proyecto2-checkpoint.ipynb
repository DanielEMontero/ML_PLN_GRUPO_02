## Datos para la predicción de género en películas

![image info](https://raw.githubusercontent.com/albahnsen/MIAD_ML_and_NLP/main/images/moviegenre.png)

En este proyecto se usará un conjunto de datos de géneros de películas. Cada observación contiene el título de una película, su año de lanzamiento, la sinopsis o plot de la película (resumen de la trama) y los géneros a los que pertenece (una película puede pertenercer a más de un género). Por ejemplo:
- Título: 'How to Be a Serial Killer'
- Plot: 'A serial killer decides to teach the secrets of his satisfying career to a video store clerk.'
- Generos: 'Comedy', 'Crime', 'Horror'

La idea es que usen estos datos para predecir la probabilidad de que una película pertenezca, dada la sinopsis, a cada uno de los géneros.

### Preprocesamiento de datos

```python
# Separación de muestras en train y test
X = dataTraining.drop(columns=['genres','year'])
y = dataTraining['genres']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
X_train
```

```python
mlb = MultiLabelBinarizer()
y_train =  mlb.fit_transform(y_train)
y_test =  mlb.transform(y_test)
```

### Modelamiento de datos

#### Red Neuronal Mono Capa

```python
vect_plot = CountVectorizer(max_features=5000)
vect_title = CountVectorizer(max_features=5000)
plot_train = vect_plot.fit_transform(X_train['plot']).toarray()
plot_test = vect_plot.transform(X_test['plot']).toarray()
title_train = vect_title.fit_transform(X_train['title']).toarray()
title_test = vect_title.transform(X_test['title']).toarray()
```

```python
X_train_mc = np.hstack((plot_train, title_train))
X_test_mc = np.hstack((plot_test, title_test))
```

```python
# Definición de dimensiones de salida, varaibles de interés
output_var = y_train.shape[1]
print(output_var, ' output variables')

# Definición de dimensiones de entrada, varaibles predictoras
dims = X_train_mc.shape[1]
print(dims, 'input variables')
```

```python
K.clear_session()

# Definición red neuronal con la función Sequential()
model = Sequential()
model.add(Dense(256, input_shape=(dims,), activation='relu'))
model.add(Dense(256, activation='relu'))          
model.add(Dense(output_var))
model.add(Activation('sigmoid'))


# Impresión de la arquitectura de la red neuronal
print(model.summary())
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy',metrics=[tf.keras.metrics.AUC(name='auc')])
```

```python
model.fit(X_train_mc, y_train, 
          verbose=1, 
          epochs=100, 
          validation_data=(X_test_mc, y_test),
          callbacks=[PlotLossesKeras()])
```

```python
y_pred_test = model.predict(X_test_mc)
roc_auc_score(y_test, y_pred_test, average='macro')
```

#### Red Neuronal Multicapa

```python
K.clear_session()

model = Sequential()
model.add(Dense(512, activation='relu', input_shape=(dims,)))
model.add(Dropout(0.3))
model.add(Dense(256, activation='relu'))
model.add(Dropout(0.3))
model.add(Dense(24, activation='sigmoid'))

print(model.summary())
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=[tf.keras.metrics.AUC(name='auc')])
early_stop=EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
```

```python
y_pred_test = model.predict(X_test_mc)
roc_auc_score(y_test, y_pred_test, average='macro')
```

```python
sorted_counts = sorted(tokenizer.word_counts.items(), key=lambda x: x[1], reverse=True)
frequencies = [count for word, count in sorted_counts]

plt.plot(frequencies[:50000])
plt.yscale('log')
plt.title("Frecuencia de las palabras (escala logarítmica)")
plt.xlabel("Rango de palabras")
plt.ylabel("Frecuencia (log)")
plt.grid(True, which="both", ls="--")
plt.show()
```

```python
# Parámetros
vocab_size = 20000
maxlen = 400

# Tokenizer para plot
tokenizer_plot = Tokenizer(num_words=vocab_size, oov_token="<OOV>")
tokenizer_plot.fit_on_texts(X_train['plot'].astype(str))
X_train_plot_seq = tokenizer_plot.texts_to_sequences(X_train['plot'].astype(str))
X_test_plot_seq = tokenizer_plot.texts_to_sequences(X_test['plot'].astype(str))
X_train_plot_pad = pad_sequences(X_train_plot_seq, maxlen=maxlen, padding='post')
X_test_plot_pad = pad_sequences(X_test_plot_seq, maxlen=maxlen, padding='post')

# Tokenizer para title
tokenizer_title = Tokenizer(num_words=vocab_size, oov_token="<OOV>")
tokenizer_title.fit_on_texts(X_train['title'].astype(str))
X_train_title_seq = tokenizer_title.texts_to_sequences(X_train['title'].astype(str))
X_test_title_seq = tokenizer_title.texts_to_sequences(X_test['title'].astype(str))
X_train_title_pad = pad_sequences(X_train_title_seq, maxlen=maxlen, padding='post')
X_test_title_pad = pad_sequences(X_test_title_seq, maxlen=maxlen, padding='post')
```

```python
X_train_lstm = [X_train_plot_pad, X_train_title_pad]
X_test_lstm = [X_test_plot_pad, X_test_title_pad]
```

```python
K.clear_session()
# Definición red neuronal con la función Sequential()
model = Sequential()

# Definición de la capa embedding
model.add(Embedding(vocab_size, 256, input_length=maxlen))
# Definición de la capa recurrente LSTM
model.add(LSTM(128))
# Definición de dropout para evitar overfitting
model.add(Dropout(0.5))
# Definición capa densa con función sigmoide para predicción binaria final
model.add(Dense(24, activation='sigmoid'))

# Definición de función de perdida.
model.compile(optimizer='adam', loss='binary_crossentropy',metrics=[tf.keras.metrics.AUC(name='auc')])

# Impresión de la arquitectura de la red neuronal
model.summary()
```

```python
model.fit(X_train_plot_pad, y_train, 
          validation_data=(X_test_plot_pad, y_test), 
          batch_size=128, 
          epochs=30, 
          verbose=1,
          callbacks=[PlotLossesKeras()])
```

```python
y_pred_test = model.predict(X_test_plot_pad)
roc_auc_score(y_test, y_pred_test, average='macro')
```
